```python
#EXPERIMENT 25

import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

gradient_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
gradient_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

gradient = cv2.subtract(gradient_x, gradient_y)
gradient = cv2.convertScaleAbs(gradient)

print("Original Image")
cv2_imshow(img)

print("Gradient Masked Image")
cv2_imshow(gradient)

cv2.imwrite("/content/Gradient_Image.jpg", gradient)

print("Image saved successfully.")
```

```python
#EXPERIMENT 26

import cv2
from google.colab.patches import cv2_imshow

# Read images
logo = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")
img = cv2.imread("/content/image.png")

# Resize logo
logo = cv2.resize(logo, (150, 150))

# Get dimensions
h_logo, w_logo, _ = logo.shape
h_img, w_img, _ = img.shape

# Center position
x = (w_img - w_logo) // 2
y = (h_img - h_logo) // 2

# Blend watermark
roi = img[y:y+h_logo, x:x+w_logo]
watermark = cv2.addWeighted(roi, 0.7, logo, 0.3, 0)

img[y:y+h_logo, x:x+w_logo] = watermark

# Display images
print("Original Image")
cv2_imshow(cv2.imread("/content/image.png"))

print("Watermarked Image")
cv2_imshow(img)

# Save image
cv2.imwrite("/content/Watermarked_Image.jpg", img)

print("Image saved successfully.")
```

```python
#EXPERIMENT 27

import cv2
from google.colab.patches import cv2_imshow

# Read images
img1 = cv2.imread("/content/image.png")
img2 = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

# Crop a portion from img1
crop = img1[50:250, 50:250]

print("Cropped Image")
cv2_imshow(crop)

# Resize img2 to match crop size
# The actual size of img1[50:250, 50:250] is (141, 113).
# cv2.resize expects (width, height) so it should be (113, 141).
img2 = cv2.resize(img2, (113, 141))

# Paste img2 into img1
img1[50:250, 50:250] = img2

print("Final Image")
cv2_imshow(img1)

# Save the output image
cv2.imwrite("/content/Copy_Paste_Image.jpg", img1)

print("Image saved successfully.")
```

```python
#EXPERIMENT 28

import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

# Convolution kernel
kernel = np.array([[-1,-1,-1],
                   [-1, 9,-1],
                   [-1,-1,-1]])

# Apply convolution
boundary = cv2.filter2D(img, -1, kernel)

# Display images
print("Original Image")
cv2_imshow(img)

print("Boundary Image")
cv2_imshow(boundary)

# Save image
cv2.imwrite("/content/Boundary_Image.jpg", boundary)

print("Image saved successfully.")
```

```python
#EXPERIMENT 40


import cv2
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

# Rectangle coordinates
start_point = (50, 50)
end_point = (250, 250)

# Draw rectangle
rect_img = img.copy()
cv2.rectangle(rect_img, start_point, end_point, (0, 255, 0), 2)

# Extract object
object_img = img[start_point[1]:end_point[1], start_point[0]:end_point[0]]

# Display images
print("Image with Rectangle")
cv2_imshow(rect_img)

print("Extracted Object")
cv2_imshow(object_img)

# Save extracted object
cv2.imwrite("/content/Extracted_Object.jpg", object_img)

print("Image saved successfully.")
```
