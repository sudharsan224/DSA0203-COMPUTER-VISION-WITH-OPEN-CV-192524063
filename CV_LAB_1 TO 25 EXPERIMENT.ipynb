```python
#EXPERIMENT 8
import cv2
from google.colab.patches import cv2_imshow

# Read the image
img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

# Get image height and width
height, width = img.shape[:2]

# Make the image bigger
bigger = cv2.resize(img, (int(width*3), int(height*3)))

# Make the image smaller
smaller = cv2.resize(img, (int(width*0.5), int(height*0.5)))

# Display images
print("Original Image")
cv2_imshow(img)

print("Bigger Image")
cv2_imshow(bigger)

print("Smaller Image")
cv2_imshow(smaller)

# Save images
cv2.imwrite("Bigger_Image.jpg", bigger)
cv2.imwrite("Smaller_Image.jpg", smaller)
```

```python
#EXPERIMENT 11
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

rows, cols = img.shape[:2]

pts1 = np.float32([[50,50], [200,50], [50,200]])
pts2 = np.float32([[10,100], [200,50], [100,250]])

matrix = cv2.getAffineTransform(pts1, pts2)

result = cv2.warpAffine(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("Affine Transformed Image")
cv2_imshow(result)

cv2.imwrite("/content/Affine_Image.jpg", result)

print("Image saved successfully")
```

```python
#EXPERIMENT 12
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

rows, cols = img.shape[:2]

pts1 = np.float32([[0,0], [cols-1,0], [0,rows-1], [cols-1,rows-1]])
pts2 = np.float32([[50,50], [cols-50,0], [0,rows-50], [cols-50,rows-50]])

matrix = cv2.getPerspectiveTransform(pts1, pts2)

result = cv2.warpPerspective(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("Perspective Transformed Image")
cv2_imshow(result)

cv2.imwrite("/content/Perspective_Image.jpg", result)

print("Image saved successfully")
```

```python
#EXPERIMENT 13
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

rows, cols = img.shape[:2]

pts1 = np.float32([[0,0], [cols-1,0], [0,rows-1], [cols-1,rows-1]])
pts2 = np.float32([[0,0], [cols-1,0], [0,int(0.7*rows)], [cols-1,int(0.7*rows)]])

matrix, _ = cv2.findHomography(pts1, pts2)

result = cv2.warpPerspective(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("Homography Transformed Image")
cv2_imshow(result)

cv2.imwrite("/content/Homography_Image.jpg", result)

print("Image saved successfully")
```

```python
#EXPERIMENT 14
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

rows, cols = img.shape[:2]

src = np.float32([[0,0], [cols-1,0], [0,rows-1], [cols-1,rows-1]])
dst = np.float32([[50,50], [cols-50,20], [20,rows-50], [cols-30,rows-30]])

matrix, _ = cv2.findHomography(src, dst)

result = cv2.warpPerspective(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("DLT Transformed Image")
cv2_imshow(result)

cv2.imwrite("/content/DLT_Transformed_Image.jpg", result)

print("Image saved successfully")
```

```python
#EXPERIMENT 19
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

edges = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

print("Original Image")
cv2_imshow(img)

print("Sobel XY Edge Detection")
cv2_imshow(cv2.convertScaleAbs(edges))

cv2.imwrite("/content/Sobel_XY.jpg", cv2.convertScaleAbs(edges))

print("Image saved successfully")
```

```python
#EXPERIMENT 23
from PIL import Image, ImageFilter
from IPython.display import display

img = Image.open("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

sharpened = img.filter(ImageFilter.UnsharpMask(radius=3, percent=200, threshold=5))

print("Original Image")
display(img)

print("Sharpened Image")
display(sharpened)

sharpened.save("/content/Unsharp_Image.jpg")

print("Image saved successfully")
```

```python
#EXPERIMENT 24
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/WhatsApp Image 2026-07-09 at 10.21.32 AM.jpeg")

blur = cv2.blur(img, (3,3))

sharpened = cv2.addWeighted(img, 2.5, blur, -1.5, 0)

print("Original Image")
cv2_imshow(img)

print("High-Boost Sharpened Image")
cv2_imshow(sharpened)

cv2.imwrite("/content/HighBoost_Image.jpg", sharpened)

print("Image saved successfully")
```
