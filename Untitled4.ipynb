```python
# ==========================================
# IMAGE REPRESENTATION USING OPENCV
# GOOGLE COLAB
# ==========================================

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# ------------------------------------------
# 1. Upload Input Image
# ------------------------------------------

uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Read image
image = cv2.imread(filename)

# OpenCV reads image in BGR format
# Convert BGR to RGB
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# ------------------------------------------
# 2. RGB Representation
# ------------------------------------------

plt.figure(figsize=(6, 5))
plt.imshow(rgb)
plt.title("RGB Image Representation")
plt.axis("off")
plt.show()


# ------------------------------------------
# 3. Grayscale Representation
# ------------------------------------------

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(6, 5))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image Representation")
plt.axis("off")
plt.show()


# ------------------------------------------
# 4. Binary Representation
# ------------------------------------------

_, binary = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)

plt.figure(figsize=(6, 5))
plt.imshow(binary, cmap="gray")
plt.title("Binary Image Representation")
plt.axis("off")
plt.show()


# ------------------------------------------
# 5. HSV Representation
# ------------------------------------------

hsv = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2HSV
)

# Display HSV image
hsv_rgb = cv2.cvtColor(
    hsv,
    cv2.COLOR_HSV2RGB
)

plt.figure(figsize=(6, 5))
plt.imshow(hsv_rgb)
plt.title("HSV Image Representation")
plt.axis("off")
plt.show()


# ------------------------------------------
# 6. Display All Representations
# ------------------------------------------

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(rgb)
plt.title("RGB")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(binary, cmap="gray")
plt.title("Binary")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(hsv_rgb)
plt.title("HSV")
plt.axis("off")

plt.tight_layout()
plt.show()


# ------------------------------------------
# 7. Display Image Information
# ------------------------------------------

print("IMAGE REPRESENTATION INFORMATION")
print("-----------------------------------")

print("Image Size:", image.shape)
print("RGB Shape:", rgb.shape)
print("Grayscale Shape:", gray.shape)
print("Binary Shape:", binary.shape)
print("HSV Shape:", hsv.shape)

print("\nPixel Value Examples:")

print("RGB Pixel:", rgb[0, 0])
print("Grayscale Pixel:", gray[0, 0])
print("HSV Pixel:", hsv[0, 0])
```

```python
# ==========================================
# OBJECT REPRESENTATION USING OPENCV
# ==========================================

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# 1. Upload an image
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# 2. Read the image
img = cv2.imread(filename)

# Convert BGR to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display input image
plt.figure(figsize=(8, 6))
plt.imshow(rgb)
plt.title("Input Image")
plt.axis("off")
plt.show()


# 3. Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# 4. Thresholding
_, binary = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY
)

# Display binary image
plt.figure(figsize=(8, 6))
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")
plt.show()


# 5. Find contours
contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

print("Number of objects:", len(contours))


# 6. Create output image
output = img.copy()

object_count = 0

for contour in contours:

    # Calculate area
    area = cv2.contourArea(contour)

    # Remove small noise
    if area < 500:
        continue

    object_count += 1

    # Calculate perimeter
    perimeter = cv2.arcLength(
        contour, True
    )

    # Bounding box
    x, y, w, h = cv2.boundingRect(contour)

    # Calculate centroid
    M = cv2.moments(contour)

    if M["m00"] != 0:
        cx = int(M["m10"] / M["m00"])
        cy = int(M["m01"] / M["m00"])
    else:
        cx = 0
        cy = 0

    # Draw contour
    cv2.drawContours(
        output,
        [contour],
        -1,
        (0, 255, 0),
        3
    )

    # Draw bounding box
    cv2.rectangle(
        output,
        (x, y),
        (x+w, y+h),
        (255, 0, 0),
        2
    )

    # Draw centroid
    cv2.circle(
        output,
        (cx, cy),
        6,
        (0, 0, 255),
        -1
    )

    # Write object number
    cv2.putText(
        output,
        "Object " + str(object_count),
        (x, y-10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (255, 0, 255),
        2
    )

    # Print properties
    print("\nObject", object_count)
    print("----------------------")
    print("Area      :", area)
    print("Perimeter :", perimeter)
    print("Centroid  :", (cx, cy))
    print("Width     :", w)
    print("Height    :", h)


# 7. Display final result
output_rgb = cv2.cvtColor(
    output,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(10, 7))
plt.imshow(output_rgb)
plt.title("Object Representation")
plt.axis("off")
plt.show()
```
